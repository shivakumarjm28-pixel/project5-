```python

# Separate input features (X) and target (y)

X = df_clean.drop(
    columns=['Booking_ID', 'booking_status']
)

y = df_clean['booking_status']

print("Input Features Shape:", X.shape)

print("Target Shape:", y.shape)
```

```python

# Identify numerical columns

numeric_features = X.select_dtypes(
    include=['int64', 'float64']
).columns.tolist()

# Identify categorical columns

categorical_features = X.select_dtypes(
    include=['object']
).columns.tolist()

print("Numerical Features:")
print(numeric_features)

print("\nCategorical Features:")
print(categorical_features)
```

```python

from sklearn.model_selection import train_test_split

# Split dataset into training and testing data

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Records:", X_train.shape[0])

print("Testing Records:", X_test.shape[0])
```

```python

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

# Numerical data processing

numeric_transformer = StandardScaler()

# Categorical data processing

categorical_transformer = OneHotEncoder(
    handle_unknown='ignore'
)

# Combine both transformations

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

print("Preprocessing setup completed!")
```

```python

from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

# Create Logistic Regression pipeline

lr_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(max_iter=1000))
])

# Train the model

lr_model.fit(X_train, y_train)

print("Logistic Regression Model Trained Successfully!")
```

```python

# Predict booking status using test data

lr_predictions = lr_model.predict(X_test)

print("First 10 Actual Values:")
print(y_test.head(10).tolist())

print("\nFirst 10 Predicted Values:")
print(lr_predictions[:10])
```

```python

from sklearn.ensemble import RandomForestClassifier

# Create Random Forest pipeline

rf_model = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ))
])

# Train the model

rf_model.fit(X_train, y_train)

print("Random Forest Model Trained Successfully!")


```

```python

# Predict booking status

rf_predictions = rf_model.predict(X_test)

print("First 10 Actual Values:")
print(y_test.head(10).tolist())

print("\nFirst 10 Random Forest Predictions:")
print(rf_predictions[:10])
```

```python

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

# Calculate evaluation metrics

print("LOGISTIC REGRESSION RESULTS")

print("Accuracy:",
      accuracy_score(y_test, lr_predictions))

print("Precision:",
      precision_score(
          y_test, lr_predictions,
          pos_label='Canceled'
      ))

print("Recall:",
      recall_score(
          y_test, lr_predictions,
          pos_label='Canceled'
      ))

print("F1 Score:",
      f1_score(
          y_test, lr_predictions,
          pos_label='Canceled'
      ))

print("\nClassification Report:")
print(classification_report(y_test, lr_predictions))
```

```python

print("RANDOM FOREST RESULTS")

print("Accuracy:",
      accuracy_score(y_test, rf_predictions))

print("Precision:",
      precision_score(
          y_test, rf_predictions,
          pos_label='Canceled'
      ))

print("Recall:",
      recall_score(
          y_test, rf_predictions,
          pos_label='Canceled'
      ))

print("F1 Score:",
      f1_score(
          y_test, rf_predictions,
          pos_label='Canceled'
      ))

print("\nClassification Report:")
print(classification_report(y_test, rf_predictions))
```

```python

# Create model comparison table

results = pd.DataFrame({
    'Model': ['Logistic Regression', 'Random Forest'],

    'Accuracy': [
        accuracy_score(y_test, lr_predictions),
        accuracy_score(y_test, rf_predictions)
    ],

    'Precision': [
        precision_score(y_test, lr_predictions,
                        pos_label='Canceled'),
        precision_score(y_test, rf_predictions,
                        pos_label='Canceled')
    ],

    'Recall': [
        recall_score(y_test, lr_predictions,
                     pos_label='Canceled'),
        recall_score(y_test, rf_predictions,
                     pos_label='Canceled')
    ],

    'F1 Score': [
        f1_score(y_test, lr_predictions,
                 pos_label='Canceled'),
        f1_score(y_test, rf_predictions,
                 pos_label='Canceled')
    ]
})

display(results.round(4))
```

```python

from sklearn.metrics import confusion_matrix

# Generate confusion matrix for Random Forest

cm = confusion_matrix(
    y_test,
    rf_predictions,
    labels=['Canceled', 'Not_Canceled']
)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Canceled', 'Not_Canceled'],
    yticklabels=['Canceled', 'Not_Canceled']
)

plt.title('Random Forest Confusion Matrix')
plt.xlabel('Predicted Status')
plt.ylabel('Actual Status')

plt.show()
```

```python

# Find the model with the highest F1 score

best_model_name = results.loc[
    results['F1 Score'].idxmax(), 'Model'
]

best_f1 = results['F1 Score'].max()

print("Best Model:", best_model_name)

print("Best F1 Score:", round(best_f1, 4))

print("\nModel Comparison:")
display(results.round(4))
```
